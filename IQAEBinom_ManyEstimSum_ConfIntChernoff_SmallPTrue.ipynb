```python
import numpy as np
from IQAEBinom import IQAEBinom
import matplotlib.pyplot as plt
```

```python
epsilon = 0.001
nEstim = 10000
alpha = 0.05 / nEstim
nShotUnit = 1
minRatio = 2
```

```python
pTrue = 0.002
errors_pTrue0002 = []
results_pTrue0002 = []
nEstim = 1000
for i in range(nEstim):
    if i % 100 == 0:
        print(i)
    result = IQAEBinom(pTrue, epsilon, alpha, nShotUnit, confint_method="chernoff", minRatio=minRatio)
    results_pTrue0002.append(result)
    errors_pTrue0002.append(result['Estimate'] - pTrue)

plt.xlabel('IdEstim')
plt.ylabel('CumErr')
ax = plt.gca()
plt.plot(np.arange(nEstim), np.array(errors_pTrue0002).cumsum())
plt.show()
```
