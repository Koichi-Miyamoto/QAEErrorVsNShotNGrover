```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
theta = 2 * np.pi / 12

ax = plt.gca()
plt.scatter(np.arange(1, 700), np.sin((2 * np.arange(1, 700) + 1) * theta) ** 2)
plt.show()
```

```python
theta = 2 * np.pi / 12

ax = plt.gca()
plt.plot(np.arange(1, 50), np.sin((2 * np.arange(1, 50) + 1) * theta) ** 2, marker="o")
plt.show()
```
