```python
import numpy as np
from IQAEBinom import IQAEBinom
import matplotlib.pyplot as plt
```

```python
epsilon = 0.001
nEstim = 1000
alpha = 0.05 / nEstim
nShotUnit = 1
minRatio = 2
```

```python
pTrue = 0.01
errors_pTrue001 = []
results_pTrue001 = []
for i in range(nEstim):
    result = IQAEBinom(pTrue, epsilon, alpha, nShotUnit, confint_method="beta", minRatio=minRatio)
    results_pTrue001.append(result)
    errors_pTrue001.append(result['Estimate'] - pTrue)

plt.xlabel('IdEstim')
plt.ylabel('CumErr')
ax = plt.gca()
plt.plot(np.arange(nEstim), np.array(errors_pTrue001).cumsum())
plt.show()
```

```python
# last nGroverの分布
plt.hist([r["nGrover"][-1] for r in results_pTrue001], range=(55, 60), bins = 5)
```
