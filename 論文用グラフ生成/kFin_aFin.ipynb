```python
import numpy as np
import matplotlib.pyplot as plt
import pickle
```

```python
p = 0.2505
with open("C:/Users/koich/Desktop/Code/QAEErrorVsNShotNGrover/output/20230830_IQAEVariousP_nShotUnit1/results_IQAE_pTrue0.25049999999999994_epsilon0001_alpha005.dat", 'rb') as f:
    results = pickle.load(f)
errs = [r["Estimate"] - p for r in results]
print(np.mean(errs), np.std(errs))
```

```python
nGs = np.array([r["nGrover"][-1] for r in results])
plt.hist(nGs, range=(0, 700), bins=35)
```

```python
thetaTrue = np.arcsin(np.sqrt(p))
ax = plt.gca()
plt.scatter(nGs, np.sin((2 * nGs + 1) * thetaTrue) ** 2, c="k", alpha=0.005)
plt.plot(np.arange(0, 700, 3), np.sin((2 * np.arange(0, 700, 3) + 1) * thetaTrue) ** 2, color='b', lw=1)
plt.plot(np.arange(1, 700, 3), np.sin((2 * np.arange(1, 700, 3) + 1) * thetaTrue) ** 2, color='r', lw=1)
plt.plot(np.arange(2, 700, 3), np.sin((2 * np.arange(2, 700, 3) + 1) * thetaTrue) ** 2, color='orange', lw=1)

ax.set_xlabel("$k_{fin}$", fontsize=12)
ax.set_ylabel("$a_{fin}$", fontsize=12)

plt.show()
```
