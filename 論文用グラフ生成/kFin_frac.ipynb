```python
import numpy as np
import matplotlib.pyplot as plt
import pickle
```

```python
p = 0.2505
with open("C:/Users/koich/Desktop/Code/QAEErrorVsNShotNGrover/output/20230830_IQAEVariousP_nShotUnit1/results_IQAE_pTrue0.25049999999999994_epsilon0001_alpha005.dat", 'rb') as f:
    results = pickle.load(f)
errs = [r["Estimate"] - p for r in results]
print(np.mean(errs), np.std(errs))
```

```python
nGs = np.array([r["nGrover"][-1] for r in results])
plt.hist(nGs, range=(0, 700), bins=35)
```

```python
plt.rcParams["font.size"] = 15

thetaTrue = np.arcsin(np.sqrt(p))
ax = plt.gca()
plt.scatter(nGs, np.modf((2 * nGs + 1) * thetaTrue / np.pi)[0], c="k", alpha=0.005)

kmod0 = np.arange(0, 700, 3)
plt.plot(kmod0, np.modf((2 * kmod0 + 1) * thetaTrue / np.pi)[0], color='b', lw=1)

kmod1 = np.arange(1, 700, 3)
plt.plot(kmod1, np.modf((2 * kmod1 + 1) * thetaTrue / np.pi)[0], color='r', lw=1)

kmod2 = np.arange(2, 700, 3)
f_kmod2 = np.modf((2 * kmod2 + 1) * thetaTrue / np.pi)[0]
i_fmax = np.argmax(f_kmod2)
plt.plot(kmod2[:i_fmax], np.modf((2 * kmod2[:i_fmax] + 1) * thetaTrue / np.pi)[0], color='orange', lw=1)

plt.plot(range(700), np.repeat(1 / 6, 700), color='b', lw=1, linestyle="dashed")
plt.plot(range(700), np.repeat(0.5, 700), color='r', lw=1, linestyle="dashed")
plt.plot(range(700), np.repeat(5 / 6, 700), color='orange', lw=1, linestyle="dashed")

plt.scatter([0.5], [0], c="k", alpha=0)

ax.set_xlabel("$k_{fin}$")
ax.set_ylabel("$f_{fin}$")

plt.show()
```
