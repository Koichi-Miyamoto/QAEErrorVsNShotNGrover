```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ptick 
import pandas as pd
```

```python
meanStdDf = pd.read_csv("C:/Users/koich/Desktop/Code/QAEErrorVsNShotNGrover/output/20230830_IQAEVariousP_nShotUnit1/summary/errMeanStd.csv")
```

```python
meanStdDf
```

```python
nEstim = 10000
ps = meanStdDf["P"]
errAvgs = meanStdDf["ErrMean"]
errStds = meanStdDf["ErrStDev"]
```

```python
ax = plt.gca()
plt.plot(ps, errAvgs, marker="o", label="Average", color="blue")
plt.plot(ps, errStds * 2 / np.sqrt(nEstim), marker=".", label="StDev * $\pm$2", color="orange")
plt.plot(ps, -errStds * 2 / np.sqrt(nEstim), marker=".", color="orange")

ax.set_xlabel("a")
ax.set_ylabel("Error")

#ax.set_ylim(1e-9, 1e-4)
#plt.yscale('log')
ax.set_ylim(-1e-4, 1e-4)
plt.legend()

ax.yaxis.set_major_formatter(ptick.ScalarFormatter(useMathText=True))
ax.ticklabel_format(style="sci", axis="y", scilimits=(-5,-5)) 

plt.show()
```

```python
plt.rcParams["font.size"] = 15

ax = plt.gca()
plt.plot(ps, np.abs(errAvgs), marker=".", label="Average", color="blue")
plt.plot(ps, errStds * 2 / np.sqrt(nEstim), marker=".", label=r"Standard Error $\times$ 2", color="orange")

ax.set_xlabel("$a$")
ax.set_ylabel("Error")

#ax.set_ylim(1e-9, 1e-4)
#plt.yscale('log')
ax.set_ylim(0, 1e-4)
plt.legend(fontsize=12)

ax.yaxis.set_major_formatter(ptick.ScalarFormatter(useMathText=True))
ax.ticklabel_format(style="sci", axis="y", scilimits=(-5,-5))
# ax.set_xticklabels(ax.get_xticklabels(), fontsize=15)
# ax.set_yticklabels(ax.get_yticklabels(), fontsize=15) 

plt.show()
```
