```python
import numpy as np
from IQAEBinom import IQAEBinom
import matplotlib.pyplot as plt
```

```python
epsilon = 0.001
nEstim = 1000
alpha = 0.05 / nEstim
nShotUnit = 1
```

```python
pTrue = 0.1
```

```python
errors = []
for i in range(nEstim):
    result = IQAEBinom(pTrue, epsilon, alpha, nShotUnit, confint_method="beta", minRatio=1.5)
    errors.append(result['Estimate'] - pTrue)

plt.xlabel('IdEstim')
plt.ylabel('CumErr')

# 描画
ax = plt.gca()
plt.plot(np.arange(nEstim), np.array(errors).cumsum())

ax.legend()
plt.show()
```

```python
minRatios = np.geomspace(1.1, 3, 100)
```

```python
resultMat = []
nEstim = 100
alpha = 0.01 / nEstim
for minRatio in minRatios:
    results = []
    for _ in range(nEstim):
        result = IQAEBinom(pTrue, epsilon, alpha, nShotUnit, confint_method="beta", minRatio=minRatio)
        results.append(result)
    resultMat.append(results)
```

```python
aveOracleCalls = [np.mean([r["TotalOracleCalls"] for r in rs]) for rs in resultMat]

plt.xlabel('minRatio')
plt.ylabel('Total oracle calls')
ax = plt.gca()
plt.plot(minRatios, aveOracleCalls)
plt.show()
```

```python
aveMaxNGrover = [np.mean([r["nGrover"][-1] for r in rs]) for rs in resultMat]
plt.xlabel('minRatio')
plt.ylabel('Max Grover Num')
ax = plt.gca()
plt.plot(minRatios, aveMaxNGrover)
plt.show()
```

```python
aveNShotLast = [np.mean([r["nShots"][-1] for r in rs]) for rs in resultMat]
plt.xlabel('minRatio')
plt.ylabel('NShot Last')
ax = plt.gca()
plt.plot(minRatios, aveNShotLast)
plt.show()
```

minRatio=1.5で多数回の推定誤差の累積和をプロットするとどうなる？

　→ 元のmodified IQAE（minRatio=3）とそんなに変わらない...結局biasの蓄積は見えない

```python
pTrue = 0.1
epsilon = 0.001
nEstim = 1000
alpha = 0.05 / nEstim
nShotUnit = 1
minRatio = 1.5
nTrial = 10

errorCumSeqs = []

for _ in range(nTrial):
    errors = []
    for _ in range(nEstim):
        result = IQAEBinom(pTrue, epsilon, alpha, nShotUnit, confint_method="beta", minRatio=minRatio)
        errors.append(result['Estimate'] - pTrue)

    errorCumSeqs.append(np.array(errors).cumsum())

plt.xlabel('IdEstim')
plt.ylabel('CumErr')

for i in range(nTrial):
    ax = plt.gca()
    plt.plot(np.arange(nEstim), errorCumSeqs[i])

ax.legend()
plt.show()
```

```python
errConcat = np.concatenate([[errorCumSeqs[0][0]], np.diff(errorCumSeqs[0])])
for errCumSeq in errorCumSeqs[1:]:
    errSeq = np.concatenate([[errCumSeq[0]], np.diff(errCumSeq)])
    errConcat = np.concatenate([errConcat, errSeq])
```
