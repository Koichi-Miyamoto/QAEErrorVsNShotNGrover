```python
import numpy as np
from IQAEBinom import IQAEBinom
import matplotlib.pyplot as plt
```

```python
epsilon = 0.001
nEstim = 1000
alpha = 0.05 / nEstim
nShotUnit = 1
```

```python
pTrue = 0.1
errors = []
for _ in range(nEstim):
    result = IQAEBinom(pTrue, epsilon, alpha, nShotUnit)
    errors.append(result['Estimate'] - pTrue)

plt.xlabel('IdEstim')
plt.ylabel('CumErr')

# 描画
ax = plt.gca()
plt.plot(np.arange(nEstim), np.array(errors).cumsum())

ax.legend()
plt.show()
```

```python
pTrue = 0.001
nTrial = 10
errorCumSeqs = []

for _ in range(nTrial):
    errors = []
    for _ in range(nEstim):
        result = IQAEBinom(pTrue, epsilon, alpha, nShotUnit)
        errors.append(result['Estimate'] - pTrue)

    errorCumSeqs.append(np.array(errors).cumsum())

plt.xlabel('IdEstim')
plt.ylabel('CumErr')

for i in range(nTrial):
    ax = plt.gca()
    plt.plot(np.arange(nEstim), errorCumSeqs[i])

ax.legend()
plt.show()
```
