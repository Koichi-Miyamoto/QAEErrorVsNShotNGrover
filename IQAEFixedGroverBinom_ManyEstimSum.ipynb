```python
import numpy as np
from IQAEFixedGroverBinom import IQAEFixedGroverBinom
import matplotlib.pyplot as plt
```

```python
epsilon = 0.001
nEstim = 1000
alpha = 0.05 / nEstim
nShotUnit = 1
```

```python
pTrue = 0.1
nGroverRatio = 2
results = []
for i in range(nEstim):
    if i % 10 == 0: 
        print(i)
    result = IQAEFixedGroverBinom(pTrue, epsilon, alpha, nShotUnit, nGroverRatio, confint_method="beta")
    results.append(result)
```

```python
result
```

```python
errors = [r["Estimate"] - pTrue for r in results]
print(np.sum(errors))
print(np.sqrt(np.var(errors) * nEstim))
```

```python
u, counts = np.unique([r["nGrover"][-1] for r in results], return_counts=True)
print(u)
print(counts)
```

```python
plt.xlabel('IdEstim')
plt.ylabel('CumErr')
ax = plt.gca()
plt.plot(np.arange(nEstim), np.array(errors).cumsum())
plt.show()
```
