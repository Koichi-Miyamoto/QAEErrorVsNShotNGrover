```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
```

```python
def FuncProb1ToAnswer(p1, nGrover, iInterval):
    
    theta0 = np.arcsin(np.sqrt(p1))
    
    if iInterval % 2 == 0:
        theta = (theta0 + np.pi * (iInterval // 2)) / (2 * nGrover + 1)  
    else:
        theta = (np.pi * (iInterval // 2 + 1) - theta0) / (2 * nGrover + 1)  
    
    return np.sin(theta) ** 2
```

```python
nGrover = 2 ** 7
```

```python
iInterval = 0

p1Grids = np.linspace(0, 1, num=1000)
answers = FuncProb1ToAnswer(p1Grids, nGrover, iInterval)

plt.xlabel('P1')
plt.ylabel('Ans')

# 描画
ax = plt.gca()
plt.plot(p1Grids, answers)
plt.show()
```
