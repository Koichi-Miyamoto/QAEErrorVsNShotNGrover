```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
from BiasCorrection import BiasCordeiroKlein, BiasTaylorExpTo4th, BiasExact
```

```python
nGrover = 100
nShot = 20
```

```python
ps = np.linspace(0.001, 1, num=1000)
biases = []
for p in ps:
    biases.append(BiasCordeiroKlein(p, nGrover, nShot))
biases = np.array(biases)

ax = plt.gca()
plt.ylim([-1e-2,1e-2])
plt.plot(ps, biases)
plt.show()
```
