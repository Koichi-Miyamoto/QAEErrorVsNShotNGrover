```python
import numpy as np
import matplotlib.pyplot as plt
import datetime
import pandas as pd
from QAESimulator import QAEAnalyticDistFunc
from MLEUtil import LikelihoodQAE
```

```python
np.random.seed(1)
```

```python
ampSqTrue = 0.3
thetaTrue = np.arcsin(np.sqrt(ampSqTrue))
print(thetaTrue)
```

NShot = 1

```python
nShot = 1
nGroverMax = 2 ** 12
```

```python
nGrovers, n1s = QAEAnalyticDistFunc(ampSqTrue, nShot, nGroverMax)
thetas = np.linspace(0, 0.5 * np.pi, num=1000)
liks = [LikelihoodQAE(theta, nGrovers, n1s, nShot) for theta in thetas]

plt.xlabel('Theta')
plt.ylabel('L')
ax = plt.gca()
plt.plot(thetas, liks, label="L")
ax.set_ylim(-500, 0)
ax.legend()
plt.show()
```

```python
plt.xlabel('Theta')
plt.ylabel('ProbDens')

# 描画
ax = plt.gca()
ps = np.exp(liks)
plt.plot(thetas, ps)
ax.legend()
plt.show()
```

```python
thetasDetail = np.linspace(0.4, 0.6, num=1000)
liksDetail = [LikelihoodQAE(theta, nGrovers, n1s, nShot) for theta in thetasDetail]

plt.xlabel('Theta')
plt.ylabel('L')
ax = plt.gca()
plt.plot(thetasDetail, liksDetail, label="L")
ax.legend()
plt.show()
```
