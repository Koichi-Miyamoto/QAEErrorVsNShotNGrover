```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import bisect 
```

```python
def ConfWidthForMarginalK(p, nShot, alpha):
    theta = np.arcsin(np.sqrt(p))
    p1Width = np.sqrt(0.5 / nShot * np.log(2 / alpha))
    k = int(0.5 * (np.arcsin(np.sqrt(1 - p1Width)) / theta - 1))
    thetau = np.arcsin(np.sqrt(np.sin((2 * k + 1) * theta) ** 2 + p1Width)) / (2 * k + 1)
    thetal = np.arcsin(np.sqrt(np.sin((2 * k + 1) * theta) ** 2 - p1Width)) / (2 * k + 1)
    return np.sin(thetau) ** 2 - np.sin(thetal) ** 2
```

```python
def MarginalPVsEsp(epsilon, nShot, alpha):
    tiny = 1e-10
    f = lambda theta: ConfWidthForMarginalK(theta, nShot, alpha) - 2 * epsilon
    return bisect(f, tiny, 1 - tiny)
```

```python
nShot = 50
alpha = 0.0001
epss = np.geomspace(1e-6, 1e-1, 100)
pMarginals = [MarginalPVsEsp(eps, nShot, alpha) for eps in epss]

ax = plt.gca()
plt.plot(epss, pMarginals / epss)
plt.xscale('log')
plt.show()
```
