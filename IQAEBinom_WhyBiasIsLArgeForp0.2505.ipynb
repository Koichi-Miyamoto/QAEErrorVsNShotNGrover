たまたま、$\sin(12\theta_{true}) \approx 0, \cos(12\theta_{true}) \approx 1$みたい

→ $\sin^2((2n_{Grover}+1)\theta_{true})=\frac{1}{2}(1-\cos((4n_{Grover}+2)\theta_{true}))$が$n_{Grover}$に関して（ほぼ）周期3の関数

→ final roundでの$p_1$がバラけないから、biasが大きくなる

```python
print(np.cos(12 * thetaTrue))
print(np.sin(12 * thetaTrue))
```

例えば、$\theta_{true}=\frac{\pi}{8}$だと、$\sin(8\theta_{true}) \approx 0, \cos(8\theta_{true}) \approx -1$だから同じ現象が見られるかも？

→　そうでもなかった

```python
p_piOver8 = np.sin(np.pi / 8) ** 2
alpha = 0.05 / nEstim
results_piOver8 = np.array([])

for i in range(nEstim):
    result = IQAEBinom(p_piOver8, epsilon, alpha, nShotUnit=1, confint_method="chernoff")
    results_piOver8 = np.append(results_piOver8, result)

errs_piOver8 = [r["Estimate"] - p_piOver8 for r in results_piOver8]
ax = plt.gca()
plt.plot(np.array(errs_piOver8).cumsum())
plt.show()
```
