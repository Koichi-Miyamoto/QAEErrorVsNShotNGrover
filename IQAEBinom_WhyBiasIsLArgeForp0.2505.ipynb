試しにuShotUnit増やす

```python
p = 0.2505
alpha = 0.05 / nEstim
results_nShotUnit50 = np.array([])

for i in range(nEstim):
    result = IQAEBinom(p, epsilon, alpha, nShotUnit=50, confint_method="chernoff")
    results_nShotUnit50 = np.append(results_nShotUnit50, result)

errs_nShotUnit50 = [r["Estimate"] - p for r in results_nShotUnit50]

ax = plt.gca()
plt.plot(np.array(errs_nShotUnit50).cumsum())
plt.show()
```

```python
p = 0.2505
alpha = 0.05 / nEstim
results_nShotUnit100 = np.array([])

for i in range(nEstim):
    result = IQAEBinom(p, epsilon, alpha, nShotUnit=100, confint_method="chernoff")
    results_nShotUnit100 = np.append(results_nShotUnit100, result)

errs_nShotUnit100 = [r["Estimate"] - p for r in results_nShotUnit100]
ax = plt.gca()
plt.plot(np.array(errs_nShotUnit100).cumsum())
plt.show()
```
